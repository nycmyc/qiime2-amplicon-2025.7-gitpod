# QIIME2 Amplicon 2025.10 - Google Colab Setup

This notebook installs QIIME2 2025.10 on Google Colab.

⚠️ **Installation takes 15-20 minutes** - be patient!

📝 **Note**: You'll need to run this setup cell every time you start a new Colab session.

````python
# QIIME2 2025.10 Installation for Google Colab
import os
import sys

QIIME2_VERSION = "2025.10"
ENV_NAME = f"qiime2-amplicon-{QIIME2_VERSION}"

print(f"Installing QIIME2 {QIIME2_VERSION} on Google Colab...")
print("This will take 15-20 minutes. Please be patient!\n")

# Install Miniconda if not present
if not os.path.exists('/opt/conda'):
    print("[1/5] Installing Miniconda...")
    !wget -q https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O /tmp/miniconda.sh
    !bash /tmp/miniconda.sh -b -p /opt/conda
    !rm /tmp/miniconda.sh
    os.environ['PATH'] = f"/opt/conda/bin:{os.environ['PATH']}"
else:
    print("[1/5] Miniconda already installed ✓")

# Download QIIME2 environment file
print(f"[2/5] Downloading QIIME2 {QIIME2_VERSION} environment file...")
env_url = f"https://raw.githubusercontent.com/qiime2/distributions/refs/heads/dev/{QIIME2_VERSION}/amplicon/released/qiime2-amplicon-ubuntu-latest-conda.yml"
!wget -q {env_url} -O /tmp/qiime2-env.yml

# Create conda environment
print(f"[3/5] Creating conda environment '{ENV_NAME}'...")
print("This is the longest step (~10-15 minutes)")
!/opt/conda/bin/conda env create -n {ENV_NAME} --file /tmp/qiime2-env.yml -y -q

# Clean up
print("[4/5] Cleaning up...")
!rm /tmp/qiime2-env.yml
!/opt/conda/bin/conda clean -afy -q

# Activate environment
print(f"[5/5] Activating environment...")
os.environ['PATH'] = f"/opt/conda/envs/{ENV_NAME}/bin:{os.environ['PATH']}"
sys.path.insert(0, f"/opt/conda/envs/{ENV_NAME}/lib/python3.9/site-packages")

# Verify installation
print("\n" + "="*50)
print("🎉 QIIME2 Installation Complete!")
print("="*50)
!/opt/conda/envs/{ENV_NAME}/bin/qiime --version
print("\nYou can now use QIIME2 commands with: !qiime <command>")
````

## Verify Installation

````python
# Verify QIIME2 is working
!qiime info
````
