## Version Verification

````python
# Verify QIIME2 version
import qiime2
import subprocess
import sys

print(f"Python version: {sys.version}")
print(f"QIIME2 version: {qiime2.__version__}")

# Assert correct version
assert qiime2.__version__.startswith('2025.10'), f"Expected QIIME2 2025.10, got {qiime2.__version__}"
print("✅ QIIME2 2025.10 verified!")
````

## Check Available Plugins

````python
# List QIIME2 plugins
result = subprocess.run(['qiime', 'info'], capture_output=True, text=True)
print(result.stdout)
````

## Import Sample Metadata

Let's load the Moving Pictures tutorial metadata.

````python
import pandas as pd
import os

# Download metadata if not already present
metadata_file = os.path.expanduser('~/tutorials/moving-pictures/sample-metadata.tsv')
if os.path.exists(metadata_file):
    metadata = pd.read_csv(metadata_file, sep='\t', index_col=0)
    print(f"Loaded metadata with {len(metadata)} samples")
    display(metadata.head())
else:
    print(f"Metadata file not found at {metadata_file}")
    print("You can download it with:")
    print('!wget "https://data.qiime2.org/2025.10/tutorials/moving-pictures/sample-metadata.tsv" -O ~/tutorials/moving-pictures/sample-metadata.tsv')
````
